# Graph RAG from scratch — `grag-ouz`

The companion to `rag-ouz`. Same document, same models, same machine — but instead of
treating the manual as a **bag of unrelated chunks**, we build a **knowledge graph** out
of it and retrieve by walking that graph.

## What was wrong with plain RAG?

In `rag-ouz`, retrieval was one operation: embed the question, return the 4 most similar
chunks. That works well for **lookup** questions:

> *"What is the underlying cause of death?"* → the chunk defining it scores highest. Done.

It fails on questions where the answer is **spread across the document** or defined by
**relationships between parts**:

| Question type | Example | Why plain RAG struggles |
|---|---|---|
| Aggregation | *"Which codes are used for HIV and where?"* | The answer is 12 chunks in 4 sections. Top-4 by similarity returns whichever 4 phrase it best — not the complete set. |
| Cross-reference | *"Which rules point to Section 4.2.3?"* | Chunks that *cite* 4.2.3 rarely contain wording similar to the question. Similarity search cannot see a citation. |
| Boundary-spanning | *"What are the steps of the selection procedure?"* | Step 3 lives in the chunk after the one that matched. Plain RAG never looks at neighbours. |

Every one of those is a **relationship** problem, and relationships are exactly what a
graph stores.

## What Graph RAG actually is

The idea in one line: **find an entry point with vectors, then expand along edges.**

```
                    ┌─ vector search ─┐
   question ───────►│  find 3 seed    │
                    │  chunks         │
                    └────────┬────────┘
                             │  seeds
                             ▼
                    ┌─────────────────────────────┐
                    │  WALK THE GRAPH             │
                    │   • same section?           │
                    │   • mentions the same code? │
                    │   • cites the same rule?    │
                    │   • adjacent in the text?   │
                    └────────┬────────────────────┘
                             │  seeds + connected chunks
                             ▼
                        prompt ──► LLM ──► answer
```

Plain RAG asks *"what looks like the question?"*
Graph RAG asks that **first**, then *"what is connected to what looks like the question?"*

## Where our graph comes from

Real Graph RAG systems often use an LLM to read every chunk and extract
`(subject, relation, object)` triples. That is slow and, with a 0.5B model on a CPU,
unreliable. We do something better suited to this document: **mine the structure that
is already there.** The ICD-10 manual gives us, for free:

- a numbered section hierarchy (`4.2` contains `4.2.3`) — from the table of contents
- ~1,500 ICD code mentions (`B20.7`, `E10`) — shared codes link chunks across sections
- explicit cross-references (*"see Section 4.2.3"*) — 47 of them
- reading order — chunk *n* precedes chunk *n+1*

That is four edge types, extracted deterministically in seconds. Step 10 shows the
LLM-extraction approach too, so you can compare.

## Roadmap

| Step | What happens |
|------|--------------|
| 1 | Read the PDF |
| 2 | Parse the table of contents into a section tree |
| 3 | Chunk the text and attach each chunk to its section |
| 4 | Extract entities: ICD codes, cross-references, chapters |
| 5 | **Build the knowledge graph** |
| 6 | Look at it — stats and a drawing |
| 7 | Embed chunks (the entry points) |
| 8 | **Graph retrieval: seed then expand** |
| 9 | Generate, and compare plain RAG vs Graph RAG head to head |
| 10 | Optional: LLM triple extraction |
| 11 | Failure modes and exercises |

**Run the cells in order.**

# Step 1 — Read the PDF

Nothing new here — this is the same extraction as `rag-ouz`. S0053 is a well-behaved
digital PDF: 252 pages, about 473,000 characters, and `pypdf` gets essentially all of it.

We keep page numbers because the graph will use them to attach chunks to sections.

````python
def extract_pages(pdf_path, max_pages=None, min_chars=20):
    """Return [(page_number, text), ...] using 1-based page numbers."""
    reader = PdfReader(str(pdf_path))
    total = len(reader.pages)
    limit = total if max_pages is None else min(total, max_pages)

    pages = []
    for i in range(limit):
        text = (reader.pages[i].extract_text() or "").strip()
        if len(text) >= min_chars:
            pages.append((i + 1, text))
    return pages, total


t0 = time.time()
pages, total_pages = extract_pages(PDF_PATH, MAX_PAGES)
print(f"{PDF_PATH.name}: {total_pages} pages in file, indexing {len(pages)}")
print(f"characters: {sum(len(t) for _, t in pages):,}")
print(f"took {time.time() - t0:.1f}s")
````

# Step 2 — Mine the section hierarchy

This is the first piece of graph structure, and it comes from headings like these,
sitting on their own line in the body text:

```
2.4.3 Blocks of categories
4.2.4 Special instructions on obvious cause (Step SP6)
```

Two useful facts follow immediately:

1. **The numbering encodes the hierarchy.** `4.2.3` is a child of `4.2`, which is a child
   of `4`. Nothing to infer — just split on the dots. That is our first edge type,
   `HAS_SUBSECTION`.

2. **The page it appears on is where that section starts.** Sort the headings by page and
   every chunk can be attributed to whichever section most recently began.

## Why not just parse the table of contents?

That was the obvious first attempt, and it **fails on this document**. Contents lines
wrap when the title is long, which strands the page number on the following line:

```
'2.2 The concept of a ‘family’ of disease and health-related '
'classifications 3'                        <-- the page number is down here
```

A line-by-line parser silently drops every wrapped entry — and those are the important
top-level sections. Top-level numbers are also written `1.` with a trailing dot, which
breaks a naive number pattern too. The result was a hierarchy missing most of its
internal nodes, with chunks from pages 70 to 100 all attributed to a single section.

Scanning the body avoids both problems and gives the **true** start page rather than one
computed from a printed-page offset.

One catch: we require at least one dot in the number (`4.1`, not `4`). Top-level headings
like *"4. Rules and guidelines..."* are repeated as a running head on every page of the
chapter, so matching them would mark a new section start on all of them. We recover the
top-level sections in the next cell instead, by inference.

````python
# A heading line: number with at least one dot, then a title, nothing else.
# Requiring the dot excludes top-level running heads (see the note above).
HEADING_RE = re.compile(r"^\s*(\d+(?:\.\d+)+)\s+(\S.{2,70}?)\s*$")


def find_sections(pages, body_starts_at=BODY_STARTS_AT):
    """Scan body pages for numbered headings -> {number: {...}}."""
    found = {}
    for page_no, text in pages:
        if page_no < body_starts_at:
            continue                       # skip cover and contents
        for line in text.split("\n"):
            m = HEADING_RE.match(line)
            if not m:
                continue
            number, title = m.group(1), m.group(2).strip()
            if not re.search(r"[A-Za-z]{3}", title):
                continue                   # titles must contain real words
            # First occurrence wins: a heading appears once, at its section start.
            found.setdefault(number, {"number": number, "title": title,
                                      "pdf_page": page_no, "synthetic": False})
    return found


def add_missing_ancestors(found):
    """A heading '4.2.1' proves '4.2' and '4' exist, even if never matched.

    Without this the hierarchy is full of holes and HAS_SUBSECTION edges cannot
    be drawn. A synthetic parent starts on the first page of its earliest child.
    """
    for number in list(found):
        parts = number.split(".")
        for i in range(1, len(parts)):
            ancestor = ".".join(parts[:i])
            child_page = found[number]["pdf_page"]
            if ancestor not in found:
                found[ancestor] = {"number": ancestor, "title": f"(part {ancestor})",
                                   "pdf_page": child_page, "synthetic": True}
            else:
                found[ancestor]["pdf_page"] = min(found[ancestor]["pdf_page"], child_page)
    return found


raw = find_sections(pages)
print(f"headings found in the body : {len(raw)}")
raw = add_missing_ancestors(raw)
print(f"after adding ancestors     : {len(raw)}")

for s in raw.values():
    s["parent"] = s["number"].rsplit(".", 1)[0] if "." in s["number"] else None
    s["depth"] = s["number"].count(".")

# Sort by page, then by depth so that when a parent and child start on the same
# page the DEEPER (more specific) one is matched last -- see section_for_page.
sections = sorted(raw.values(), key=lambda s: (s["pdf_page"], s["depth"]))

print()
for s in sections[:14]:
    mark = "~" if s["synthetic"] else " "
    print(f" {mark}{'   ' * s['depth']}{s['number']:9} {s['title'][:44]:46} p.{s['pdf_page']}")
print("\n(~ = inferred parent, never appeared as a heading on its own)")
````

````python
# Print the hierarchy as a tree, to confirm the parent links are sane.
by_parent = defaultdict(list)
for s in sections:
    by_parent[s["parent"]].append(s)


def print_tree(parent=None, depth=0, max_depth=2):
    for s in sorted(by_parent[parent], key=lambda x: [int(p) for p in x["number"].split(".")]):
        print("    " * depth + f"{s['number']}  {s['title'][:52]}")
        if depth < max_depth:
            print_tree(s["number"], depth + 1, max_depth)


print("SECTION TREE (top 3 levels)\n")
print_tree()
````

# Step 3 — Chunk, and attach each chunk to a section

Chunking is unchanged from `rag-ouz` — 800 characters, 120 overlap, cutting on sentence
boundaries where possible.

What is new is the last line of the loop: every chunk records **which section it belongs
to**. We find it by walking the section list and taking the last one that started on or
before the chunk's page. That single field becomes the `CONTAINS` edge, and it is what
lets retrieval later say *"give me the rest of this rule, not just the sentence that
matched."*

````python
def clean_text(text):
    """Undo PDF layout artefacts that carry no meaning."""
    text = text.replace("­", "")
    text = re.sub(r"-\n(\w)", r"\1", text)     # re-join hyphen-split words
    text = re.sub(r"[ \t]+", " ", text)
    text = re.sub(r"\n{3,}", "\n\n", text)
    return text.strip()


def chunk_text(text, size=CHUNK_SIZE, overlap=CHUNK_OVERLAP):
    """Overlapping chunks, preferring to end on a sentence boundary."""
    text = clean_text(text)
    if not text:
        return []
    if len(text) <= size:
        return [text]

    chunks, start = [], 0
    while start < len(text):
        end = min(start + size, len(text))
        window = text[start:end]
        if end < len(text):
            cut = max(window.rfind(". "), window.rfind("\n"))
            if cut > size * 0.5:
                window = window[:cut + 1]
                end = start + len(window)
        piece = window.strip()
        if piece:
            chunks.append(piece)
        if end >= len(text):
            break
        start = end - overlap
    return chunks


def section_for_page(page_no):
    """The section a given PDF page falls inside: the last one to have started."""
    current = None
    for s in sections:                       # already sorted by pdf_page
        if s["pdf_page"] <= page_no:
            current = s
        else:
            break
    return current


records = []
for page_no, page_text in pages:
    sec = section_for_page(page_no)
    for j, piece in enumerate(chunk_text(page_text)):
        records.append({
            "id": f"p{page_no}-c{j}",
            "text": piece,
            "page": page_no,
            "section": sec["number"] if sec else None,
            "section_title": sec["title"] if sec else "(front matter)",
        })

print(f"Chunks: {len(records)}")
attached = sum(1 for r in records if r["section"])
print(f"Attached to a section: {attached}/{len(records)}")
print("\nExample:")
r0 = records[len(records) // 2]
print(f"  id      : {r0['id']}")
print(f"  page    : {r0['page']}")
print(f"  section : {r0['section']} - {r0['section_title']}")
print(f"  text    : {r0['text'][:220]}...")
````

# Step 4 — Extract entities

Now the other three edge types. All three are regular expressions — no LLM, no
guesswork, and it runs in well under a second.

**ICD codes** (`B20.7`, `E10`, `I21`) — a letter, two digits, optionally a dot and more
digits. When two chunks in different sections both mention `B20.7`, they are talking about
the same thing, and that is a link worth having.

**Cross-references** (*"see Section 4.2.3"*) — the author explicitly telling us two parts
of the manual are related. This is the most valuable edge in the whole graph, because it
is the one plain vector search is completely blind to: a chunk saying *"apply the rule in
Section 4.2.3"* shares almost no vocabulary with Section 4.2.3 itself.

**Chapter references** (*"Chapter XX"*) — coarser, but useful for grouping.

A note on precision: `A00` inside the phrase "categories A00–T98" is a range, not really a
single code. We accept that noise. Cheap-and-slightly-wrong beats slow-and-perfect for a
first pass, and Step 11 asks you to tighten it.

````python
# Letter + 2 digits, optional .digits  ->  E10, B20.7, I21.9
CODE_RE = re.compile(r"\b([A-Z]\d{2}(?:\.\d+)?)\b")
# "Section 4.2.3" / "section 4.2"
XREF_RE = re.compile(r"[Ss]ection\s+(\d+(?:\.\d+)*)")
# "Chapter XX" in roman numerals
CHAPTER_RE = re.compile(r"\bChapter\s+([IVXLC]+)\b")

valid_sections = {s["number"] for s in sections}

for rec in records:
    text = rec["text"]
    rec["codes"] = sorted(set(CODE_RE.findall(text)))
    # only keep cross-references that point at a section we actually know
    rec["refs"] = sorted({x for x in XREF_RE.findall(text) if x in valid_sections})
    rec["chapters"] = sorted(set(CHAPTER_RE.findall(text)))

code_counts = Counter(c for r in records for c in r["codes"])
ref_counts = Counter(x for r in records for x in r["refs"])

print(f"chunks with >=1 code       : {sum(1 for r in records if r['codes'])}/{len(records)}")
print(f"distinct codes             : {len(code_counts)}")
print(f"codes in 2+ chunks (links!) : {sum(1 for c, n in code_counts.items() if n > 1)}")
print(f"cross-references resolved  : {sum(ref_counts.values())} -> {len(ref_counts)} targets")

print("\nMost-mentioned codes:", code_counts.most_common(8))
print("Most-cited sections  :", ref_counts.most_common(6))
````

# Step 5 — Build the knowledge graph

A graph is just **nodes** and **edges**. Ours has four kinds of node and five kinds of
edge:

### Nodes
| Prefix | Kind | Example |
|--------|------|---------|
| `sec:` | a numbered section | `sec:4.2.3` |
| `chunk:` | ~800 characters of text | `chunk:p51-c0` |
| `code:` | an ICD code | `code:B20.7` |
| `chap:` | an ICD chapter | `chap:XX` |

### Edges
| Edge | From → To | Meaning |
|------|-----------|---------|
| `HAS_SUBSECTION` | section → section | `4.2` contains `4.2.3` |
| `CONTAINS` | section → chunk | this text belongs to that section |
| `NEXT` | chunk → chunk | reading order |
| `MENTIONS_CODE` | chunk → code | this text discusses `B20.7` |
| `REFERENCES` | chunk → section | this text cites that section |

The payoff is **indirect connection**. Two chunks 40 pages apart, sharing no vocabulary,
are two hops apart if they mention the same code:

```
chunk:p71-c2 ──MENTIONS_CODE──► code:B20.7 ◄──MENTIONS_CODE── chunk:p88-c1
```

Vector search will never put those two together. The graph does it structurally.

````python
# Node id helpers — a readable prefix makes the graph inspectable by eye.
def sec_id(number):  return f"sec:{number}"
def chunk_id(cid):   return f"chunk:{cid}"
def code_id(code):   return f"code:{code}"
def chap_id(ch):     return f"chap:{ch}"


# MultiDiGraph: directed (edges mean different things each way) and "multi"
# so two nodes can be joined by more than one kind of edge.
G = nx.MultiDiGraph()

# --- 1. sections, and the hierarchy implied by their numbering ----------
for s in sections:
    G.add_node(sec_id(s["number"]), kind="section",
               label=f"{s['number']} {s['title']}", page=s["pdf_page"])

for s in sections:
    if s["parent"] and sec_id(s["parent"]) in G:
        G.add_edge(sec_id(s["parent"]), sec_id(s["number"]), kind="HAS_SUBSECTION")

# --- 2. chunks, attached to their section -------------------------------
for rec in records:
    G.add_node(chunk_id(rec["id"]), kind="chunk", label=rec["id"],
               page=rec["page"], text=rec["text"], section=rec["section"])
    if rec["section"] and sec_id(rec["section"]) in G:
        G.add_edge(sec_id(rec["section"]), chunk_id(rec["id"]), kind="CONTAINS")

# --- 3. reading order ---------------------------------------------------
for a, b in zip(records, records[1:]):
    G.add_edge(chunk_id(a["id"]), chunk_id(b["id"]), kind="NEXT")

# --- 4. codes -----------------------------------------------------------
for rec in records:
    for c in rec["codes"]:
        if code_id(c) not in G:
            G.add_node(code_id(c), kind="code", label=c)
        G.add_edge(chunk_id(rec["id"]), code_id(c), kind="MENTIONS_CODE")

# --- 5. cross-references and chapters -----------------------------------
for rec in records:
    for target in rec["refs"]:
        G.add_edge(chunk_id(rec["id"]), sec_id(target), kind="REFERENCES")
    for ch in rec["chapters"]:
        if chap_id(ch) not in G:
            G.add_node(chap_id(ch), kind="chapter", label=f"Chapter {ch}")
        G.add_edge(chunk_id(rec["id"]), chap_id(ch), kind="MENTIONS_CHAPTER")

print(f"GRAPH BUILT: {G.number_of_nodes():,} nodes, {G.number_of_edges():,} edges")
````

````python
# What is actually in there?
node_kinds = Counter(d["kind"] for _, d in G.nodes(data=True))
edge_kinds = Counter(d["kind"] for _, _, d in G.edges(data=True))

print("NODES")
for k, n in node_kinds.most_common():
    print(f"   {k:10} {n:6,}")
print("\nEDGES")
for k, n in edge_kinds.most_common():
    print(f"   {k:16} {n:6,}")

# The busiest entity nodes: these are the hubs that connect distant chunks.
print("\nMost connected entities (the natural bridges in this document):")
ents = [(n, G.degree(n)) for n, d in G.nodes(data=True) if d["kind"] in ("code", "chapter")]
for n, deg in sorted(ents, key=lambda x: -x[1])[:10]:
    print(f"   {n:16} appears in {deg:3} chunks")
````

````python
def neighbours(node, kinds=None):
    """All nodes joined to `node`, following edges in BOTH directions.

    Direction matters when building the graph (a section CONTAINS a chunk, not
    the reverse) but not when exploring it — we want everything adjacent.
    """
    out = set()
    for _, v, d in G.out_edges(node, data=True):
        if kinds is None or d["kind"] in kinds:
            out.add(v)
    for u, _, d in G.in_edges(node, data=True):
        if kinds is None or d["kind"] in kinds:
            out.add(u)
    return out


def describe_node(node):
    """Print a node and everything it touches, grouped by edge type."""
    data = G.nodes[node]
    print(f"NODE {node}  [{data['kind']}]")
    print(f"   {data.get('label', '')[:70]}")
    grouped = defaultdict(list)
    for _, v, d in G.out_edges(node, data=True):
        grouped[d["kind"] + " ->"].append(v)
    for u, _, d in G.in_edges(node, data=True):
        grouped["<- " + d["kind"]].append(u)
    for kind, nodes in grouped.items():
        print(f"   {kind:22} {len(nodes)}: {', '.join(sorted(nodes)[:6])}"
              + (" ..." if len(nodes) > 6 else ""))


# Look at the most-cited section — the one the manual keeps pointing back to.
most_cited = ref_counts.most_common(1)[0][0] if ref_counts else sections[0]["number"]
describe_node(sec_id(most_cited))

print("\n" + "-" * 70)
# And the most-mentioned ICD code.
top_code = code_counts.most_common(1)[0][0]
describe_node(code_id(top_code))
````

# Step 6 — Look at the graph

Statistics are abstract; a picture is not. Below we draw the **ego network** of one ICD
code — the code itself, every chunk that mentions it, and the sections those chunks sit
in.

If the code appears in chunks from several different sections, you are looking at exactly
the connection plain RAG cannot make.

````python
def draw_ego(center, radius=2, max_nodes=45):
    """Draw the neighbourhood around one node, coloured by node kind."""
    ego = nx.ego_graph(G.to_undirected(as_view=False), center, radius=radius)
    if ego.number_of_nodes() > max_nodes:
        keep = [center] + sorted(
            (n for n in ego.nodes if n != center),
            key=lambda n: -G.degree(n))[:max_nodes - 1]
        ego = ego.subgraph(keep)

    colours = {"code": "#e07a5f", "chunk": "#81b29a",
               "section": "#3d5a80", "chapter": "#f2cc8f"}
    node_colours = [colours.get(G.nodes[n]["kind"], "#cccccc") for n in ego.nodes]
    sizes = [900 if n == center else 260 for n in ego.nodes]

    # Short labels: full chunk ids would make the picture unreadable.
    labels = {}
    for n in ego.nodes:
        d = G.nodes[n]
        labels[n] = d["label"][:16] if d["kind"] in ("code", "chapter") else \
                    (d.get("section") or d["label"])[:12] if d["kind"] == "chunk" else \
                    n.replace("sec:", "S")

    plt.figure(figsize=(12, 8))
    pos = nx.spring_layout(ego, seed=42, k=0.55)   # seed -> same layout every run
    nx.draw_networkx_edges(ego, pos, alpha=0.25, width=0.8)
    nx.draw_networkx_nodes(ego, pos, node_color=node_colours, node_size=sizes, alpha=0.9)
    nx.draw_networkx_labels(ego, pos, labels, font_size=7)
    plt.title(f"Neighbourhood of {center}  "
              f"({ego.number_of_nodes()} nodes)  "
              f"orange=code  green=chunk  blue=section  yellow=chapter")
    plt.axis("off")
    plt.tight_layout()
    plt.show()


draw_ego(code_id(top_code), radius=2)

# Which sections does this one code reach into?
touched = {G.nodes[c].get("section") for c in neighbours(code_id(top_code), {"MENTIONS_CODE"})}
print(f"Code {top_code} appears in sections: {sorted(x for x in touched if x)}")
````

# Step 7 — Embeddings and the vector index

The graph gives us structure, but we still need a way in. You cannot start walking a
graph until you know **which node to start from**, and a user's question is prose, not a
node id.

So we embed the chunks exactly as in `rag-ouz`. Vector search finds the **entry points**;
the graph does the rest. Same model, same settings — so any difference in the final
answers comes from the graph, not from a different embedder.

````python
from sentence_transformers import SentenceTransformer

print("Loading bge-m3 from local cache...")
t0 = time.time()
embedder = SentenceTransformer(EMBED_MODEL, device=DEVICE)
embedder.max_seq_length = 512          # chunks are ~200 tokens; big CPU saving

# The method was renamed in sentence-transformers 5; support both names.
get_dim = getattr(embedder, "get_embedding_dimension", None) \
          or embedder.get_sentence_embedding_dimension
print(f"loaded in {time.time() - t0:.1f}s, dim={get_dim()}")
````

````python
# The slow cell. ~100 pages -> ~280 chunks -> roughly 3 minutes on a CPU.
texts = [r["text"] for r in records]
print(f"Embedding {len(texts)} chunks...")

t0 = time.time()
vectors = embedder.encode(texts, batch_size=8, normalize_embeddings=True,
                          show_progress_bar=True)
print(f"\ndone in {time.time() - t0:.1f}s -> {vectors.shape}")
````

````python
client = chromadb.PersistentClient(path=str(CHROMA_DIR))

# Replace, never append — re-running must not duplicate chunks.
try:
    client.delete_collection(COLLECTION_NAME)
    print("deleted previous collection")
except Exception:
    print("no previous collection (first run)")

try:
    collection = client.create_collection(
        name=COLLECTION_NAME, configuration={"hnsw": {"space": "cosine"}})
except Exception:                       # older chromadb
    collection = client.create_collection(
        name=COLLECTION_NAME, metadata={"hnsw:space": "cosine"})

BATCH = 200
for i in range(0, len(records), BATCH):
    part = records[i:i + BATCH]
    collection.add(
        ids       =[r["id"] for r in part],
        documents =[r["text"] for r in part],
        # cid is the bridge: it turns a search result back into a graph node.
        metadatas =[{"page": r["page"],
                     "section": r["section"] or "",
                     "cid": r["id"]} for r in part],
        embeddings=vectors[i:i + BATCH].tolist(),
    )

print(f"indexed {collection.count()} chunks")
````

# Step 8 — Retrieval: plain versus graph

Here is the heart of the notebook. Two functions, same question, different strategy.

### `vector_search` — the `rag-ouz` baseline
Embed the question, return the top *k* by cosine similarity. One step, no structure.

### `graph_search` — seed and expand
1. **Seed.** `vector_search` with a *smaller* k (3 instead of 4) — we deliberately hold
   back budget for the expansion.
2. **Expand.** For each seed chunk, follow its edges one hop:
   - through a **shared ICD code** → chunks elsewhere discussing the same code
   - through its **section** → sibling chunks in the same rule
   - through a **cross-reference** → the section this chunk points at
   - through **`NEXT`** → the chunk immediately before or after
3. **Score.** A neighbour inherits `HOP_DECAY` (0.55) of the score of the seed that
   reached it, so graph results never outrank a genuine direct match.
4. **Trim.** Keep the best `GRAPH_BUDGET` chunks overall.

### Three rules that stop it exploding

**Hub suppression** (`MAX_HUB_DEGREE`). If a code appears in 60 chunks, it is a stopword,
not a link — expanding through it would drag in a third of the document. We skip any
entity connected to more chunks than the threshold.

**Per-entity cap** (`MAX_PER_ENTITY`). Even a legitimate entity contributes at most 3
chunks, so one popular code cannot crowd out every other path.

**Rarest entity first.** A chunk usually shares several entities with its neighbours, and
the budget only stretches to a few. We spend it on the **least common** ones: a code
appearing in 3 chunks says far more about a connection than one appearing in 30. This is
the same intuition as IDF in keyword search. It also makes the walk **deterministic** —
iterating a Python set gives an arbitrary order, so without sorting the same question
could return different chunks on different runs.

Without these three, a graph walk degenerates into "return everything", which is slower
than plain RAG *and* worse. Naive expansion is the classic Graph RAG failure.

Every returned chunk carries a **`via`** field explaining why it is there. Read it —
being able to see the reasoning is most of the value of building this.

````python
def vector_search(question, k=TOP_K):
    """Plain RAG retrieval: pure cosine similarity, no structure. The baseline."""
    qvec = embedder.encode([question], normalize_embeddings=True)
    res = collection.query(query_embeddings=qvec.tolist(), n_results=k,
                           include=["documents", "metadatas", "distances"])
    hits = []
    for text, meta, dist in zip(res["documents"][0], res["metadatas"][0], res["distances"][0]):
        hits.append({
            "cid": meta["cid"],
            "text": text,
            "page": meta["page"],
            "section": meta["section"],
            "score": 1.0 - dist,        # cosine distance -> similarity
            "via": "vector match",
        })
    return hits


def graph_search(question, seed_k=SEED_K, budget=GRAPH_BUDGET):
    """Graph RAG retrieval: vector search for entry points, then walk the edges."""
    seeds = vector_search(question, seed_k)

    # chunk node -> [score, explanation]
    found = {chunk_id(h["cid"]): [h["score"], h["via"]] for h in seeds}
    seed_nodes = list(found)

    for node in seed_nodes:
        base = found[node][0]

        # --- hop A: through entity nodes (codes, sections, chapters) -----
        # RAREST FIRST. A code in 3 chunks is a precise signal; one in 30 is
        # nearly a stopword. Sorting by degree spends the budget on the most
        # informative links -- the same intuition as IDF in keyword search.
        # Sorting also makes the walk deterministic: iterating a Python set
        # directly gives an arbitrary order, so results would drift run to run.
        entities = sorted(
            neighbours(node, {"MENTIONS_CODE", "REFERENCES",
                              "CONTAINS", "MENTIONS_CHAPTER"}),
            key=lambda e: (G.degree(e), e),
        )

        for ent in entities:
            # Hub suppression: an entity in dozens of chunks carries no signal.
            if G.degree(ent) > MAX_HUB_DEGREE:
                continue

            taken = 0
            for other in sorted(neighbours(ent, {"MENTIONS_CODE", "REFERENCES",
                                                 "CONTAINS", "MENTIONS_CHAPTER"})):
                if taken >= MAX_PER_ENTITY:
                    break
                if other in found or G.nodes[other]["kind"] != "chunk":
                    continue
                label = G.nodes[ent].get("label", ent)
                kind = G.nodes[ent]["kind"]
                found[other] = [base * HOP_DECAY, f"shares {kind} '{label[:38]}'"]
                taken += 1

        # --- hop B: reading-order neighbours ----------------------------
        # Cheap fix for answers that straddle a chunk boundary.
        for other in neighbours(node, {"NEXT"}):
            if other not in found:
                found[other] = [base * HOP_DECAY * 0.9, "adjacent in reading order"]

    ranked = sorted(found.items(), key=lambda kv: -kv[1][0])[:budget]

    hits = []
    for node, (score, why) in ranked:
        d = G.nodes[node]
        hits.append({
            "cid": d["label"],
            "text": d["text"],
            "page": d["page"],
            "section": d.get("section") or "",
            "score": score,
            "via": why,
        })
    return hits


print("vector_search() and graph_search() ready.")
````

````python
def compare_retrieval(question):
    """Show what each strategy retrieves, side by side."""
    print("=" * 78)
    print("QUESTION:", question)
    print("=" * 78)

    for name, hits in (("PLAIN RAG  (vector only)", vector_search(question, TOP_K)),
                       ("GRAPH RAG  (seed + expand)", graph_search(question))):
        print(f"\n--- {name} ---")
        for i, h in enumerate(hits, 1):
            sec = h["section"] or "-"
            print(f" {i}. p.{h['page']:<4} sec {sec:<8} score {h['score']:.3f}   via {h['via']}")

    v = {h["cid"] for h in vector_search(question, TOP_K)}
    g = {h["cid"] for h in graph_search(question)}
    print(f"\n  chunks only Graph RAG found: {sorted(g - v)}")
    print(f"  sections covered — plain: {sorted({h['section'] for h in vector_search(question, TOP_K) if h['section']})}")
    print(f"  sections covered — graph: {sorted({h['section'] for h in graph_search(question) if h['section']})}")


# An aggregation question: the answer is spread over several sections.
compare_retrieval("Which codes are used for HIV disease?")
````

````python
# A cross-reference question. Plain RAG cannot see citations at all — the chunks
# that POINT AT a section share almost no wording with the section itself.
compare_retrieval(f"What does section {most_cited} say about selecting the cause of death?")
````

## A query plain RAG cannot express at all

Everything above still *started* with a vector search. But once the graph exists, some
questions can be answered by structure alone — no embeddings, no LLM, no similarity.

*"Which parts of the manual cite Section 4.2.3?"* is one of them. It is not a question
about wording, so there is nothing for a vector search to match on. It is a question
about **edges**, and the answer is exact rather than ranked: follow `REFERENCES` edges
backwards and you have it.

This is the clearest illustration of what a graph buys you. Not better similarity —
a different *kind* of question.

````python
def cited_by(section_number, preview=110):
    """Which chunks cite this section? Walks REFERENCES edges backwards.

    No vectors and no LLM — this is a pure structural lookup, and the result is
    exact rather than a ranked guess.
    """
    node = sec_id(section_number)
    if node not in G:
        print(f"section {section_number} is not in the graph")
        return []

    citing = [u for u, _, d in G.in_edges(node, data=True) if d["kind"] == "REFERENCES"]
    label = G.nodes[node].get("label", node)
    print(f"Chunks citing section {section_number} — {label[:56]}")
    print(f"{len(citing)} citation(s)\n")

    for c in sorted(set(citing)):
        d = G.nodes[c]
        text = " ".join(d["text"].split())
        print(f"  p.{d['page']:<4} sec {d.get('section') or '-':<8} {text[:preview]}...")
    return citing


_ = cited_by(most_cited)

# Contrast: what would plain vector search return for the same intent?
print("\n" + "-" * 78)
print("Plain RAG, asked the same thing in words:\n")
for h in vector_search(f"which rules refer to section {most_cited}?", 3):
    print(f"  p.{h['page']:<4} sec {h['section'] or '-':<8} score {h['score']:.3f}")
print("\n...it returns whatever WORDS look similar. It has no notion of a citation.")
````

# Step 9 — Generation, and the head-to-head

Same model and same prompt for both strategies, so the only variable is **what went into
the context**. That is the whole experiment.

The prompt keeps the three grounding rules from `rag-ouz` — answer only from context,
cite pages, admit ignorance — and adds the section number to each passage, since the
graph knows it and it makes citations more precise.

````python
from transformers import AutoTokenizer, AutoModelForCausalLM

print("Loading Qwen2.5-0.5B-Instruct...")
t0 = time.time()
tokenizer = AutoTokenizer.from_pretrained(LLM_MODEL)
# No dtype argument: float32 is the CPU default and passing torch_dtype/dtype
# is the most common source of transformers version errors.
llm = AutoModelForCausalLM.from_pretrained(LLM_MODEL)
llm.to(DEVICE)
llm.eval()
print(f"loaded in {time.time() - t0:.1f}s")
````

````python
SYSTEM_PROMPT = (
    "You are a medical coding assistant. Answer the question using ONLY the "
    "context passages provided. Cite the page number for each fact, like [p.51]. "
    "If the context does not contain the answer, reply exactly: "
    "\"I cannot answer this from the provided documents.\" Never use outside knowledge."
)


def build_prompt(question, hits):
    """System rules + retrieved passages (labelled with page AND section) + question."""
    context = "\n\n".join(
        f"[p.{h['page']}" + (f", section {h['section']}" if h["section"] else "") + f"]\n{h['text']}"
        for h in hits
    )
    messages = [
        {"role": "system", "content": SYSTEM_PROMPT},
        {"role": "user", "content": f"Context passages:\n\n{context}\n\nQuestion: {question}"},
    ]
    # The tokenizer ships the model's own chat template — never hand-roll this.
    return tokenizer.apply_chat_template(messages, tokenize=False, add_generation_prompt=True)


def generate(prompt, max_new_tokens=300):
    """Run the model, return only the newly generated tokens."""
    inputs = tokenizer(prompt, return_tensors="pt").to(DEVICE)
    with torch.no_grad():
        out = llm.generate(
            **inputs,
            max_new_tokens=max_new_tokens,
            do_sample=False,            # greedy — deterministic, right for extraction
            pad_token_id=tokenizer.eos_token_id,
        )
    return tokenizer.decode(out[0][inputs["input_ids"].shape[1]:],
                            skip_special_tokens=True).strip()


def ask(question, mode="graph", show_context=False):
    """Full pipeline. mode='graph' walks the graph, mode='plain' is the baseline."""
    t0 = time.time()
    hits = graph_search(question) if mode == "graph" else vector_search(question, TOP_K)
    answer = generate(build_prompt(question, hits))

    print(f"[{mode.upper()}] {question}")
    print("-" * 78)
    print(answer)
    print("\nsources:")
    for h in hits:
        print(f"   p.{h['page']:<4} sec {h['section'] or '-':<8} {h['score']:.3f}  ({h['via']})")
    print(f"({time.time() - t0:.1f}s)\n")

    if show_context:
        print("CONTEXT SENT TO THE MODEL:")
        for h in hits:
            print(f"  [p.{h['page']}] {h['text'][:240]}...\n")
    return {"question": question, "answer": answer, "hits": hits}


print("ask() ready.")
````

````python
def head_to_head(question, show_context=False):
    """Run the same question through both strategies and print both answers."""
    print("#" * 78)
    print("#", question)
    print("#" * 78 + "\n")
    plain = ask(question, mode="plain", show_context=show_context)
    graph = ask(question, mode="graph", show_context=show_context)

    p_secs = sorted({h["section"] for h in plain["hits"] if h["section"]})
    g_secs = sorted({h["section"] for h in graph["hits"] if h["section"]})
    print(f"sections in context — plain: {p_secs}")
    print(f"sections in context — graph: {g_secs}")
    return plain, graph


# Aggregation: the codes for HIV are scattered across several rules.
_ = head_to_head("Which ICD codes are used for HIV disease?")
````

````python
# A plain lookup question, where one passage holds the whole answer.
# Expect BOTH strategies to answer it well and the graph to add nothing:
# expansion stays inside the same section. Knowing when Graph RAG does NOT
# help is as useful as knowing when it does.
_ = head_to_head("What is the purpose and applicability of the ICD?")
````

````python
# Out of scope: both strategies should refuse. Adding more context via the graph
# must not make the model more willing to invent an answer.
_ = ask("What is the best recipe for chocolate cake?", mode="graph")
````

# Step 10 — The other way to build a graph (optional)

Everything so far mined structure the document already had. That worked because the
ICD-10 manual is numbered, cross-referenced and full of codes.

Most documents are not. For a folder of meeting notes or contracts there is no table of
contents to parse, so the usual approach — and what Microsoft's GraphRAG does — is to
**have an LLM read every chunk and extract triples**:

```
"HIV disease resulting in tuberculosis is coded B20.0"
      ↓
("HIV disease", "RESULTS_IN", "tuberculosis")
("tuberculosis due to HIV", "HAS_CODE", "B20.0")
```

Then the entities become nodes and the relations become edges — same graph, different
source.

### Be realistic about the cost

**It is slow.** One LLM call per chunk. Our 280 chunks at ~20 s each on a CPU is over
90 minutes, versus under a second for the regex pass.

**Qwen2.5-0.5B is not good at it.** Reliable triple extraction needs solid instruction
following and valid JSON. A 0.5B model produces malformed output, invents relations, and
misses obvious ones. Expect roughly half of what you get to be unusable.

**It is not deterministic.** Re-run it and the graph changes, which makes debugging
retrieval genuinely painful.

So the cell below is **off by default** and capped at a handful of chunks. Turn it on to
see the shape of the technique and to feel *why* structural extraction is preferable when
the structure exists. On a 7B+ model the trade-off flips.

````python
# ---------------------------------------------------------------------------
# OPTIONAL: LLM-based triple extraction. Slow, and unreliable at this model size.
# ---------------------------------------------------------------------------
RUN_LLM_EXTRACTION = False    # <-- set True to try it
EXTRACT_N_CHUNKS = 3          # keep tiny: ~20 seconds per chunk on CPU

EXTRACT_PROMPT = """Extract factual relationships from the medical text below.
Reply with ONLY a JSON array, no other words. Each item must be:
{"subject": "...", "relation": "...", "object": "..."}
Use at most 5 items. If there are no clear relationships, reply [].

Text:
"""


def extract_triples(text):
    """Ask the LLM for triples and parse them defensively."""
    messages = [{"role": "user", "content": EXTRACT_PROMPT + text[:900]}]
    prompt = tokenizer.apply_chat_template(messages, tokenize=False, add_generation_prompt=True)
    raw = generate(prompt, max_new_tokens=220)

    # The model often wraps JSON in prose or a code fence. Grab the array.
    match = re.search(r"\[.*\]", raw, re.S)
    if not match:
        return [], raw
    try:
        data = json.loads(match.group(0))
    except json.JSONDecodeError:
        return [], raw

    triples = [
        (str(d["subject"]).strip(), str(d["relation"]).strip(), str(d["object"]).strip())
        for d in data
        if isinstance(d, dict) and {"subject", "relation", "object"} <= set(d)
    ]
    return triples, raw


if RUN_LLM_EXTRACTION:
    G_llm = nx.MultiDiGraph()
    for rec in records[:EXTRACT_N_CHUNKS]:
        t0 = time.time()
        triples, raw = extract_triples(rec["text"])
        print(f"\nchunk {rec['id']} ({time.time() - t0:.1f}s) -> {len(triples)} triples")
        if not triples:
            print("   parse failed. raw output was:", raw[:180])
        for s, r, o in triples:
            print(f"   ({s}) --{r}--> ({o})")
            G_llm.add_edge(s, o, kind=r, source=rec["id"])
    print(f"\nLLM graph: {G_llm.number_of_nodes()} nodes, {G_llm.number_of_edges()} edges")
    print("Compare with the structural graph:",
          f"{G.number_of_nodes():,} nodes, {G.number_of_edges():,} edges — built in under a second.")
else:
    print("LLM extraction is OFF.")
    print(f"For reference, the structural graph has {G.number_of_nodes():,} nodes "
          f"and {G.number_of_edges():,} edges, and took under a second to build.")
````

# Step 11 — When Graph RAG helps, when it hurts, and what to try

## Be honest about the results

You have probably noticed that on the two head-to-head questions, both strategies produced
**almost the same answer**. That is the correct outcome, and worth sitting with: Graph RAG
is not a general upgrade. It is a fix for a specific class of question, and the honest
result on an easy lookup is "no difference."

**Where it clearly wins is `cited_by()`** — the pure structural query in Step 8. Asking
*"which parts of the manual cite Section 4.2.3?"* returned 8 exact citations spanning
sections 4.1.3 through 4.2.5, one of which literally reads *"according to Section 4.2.3"*.
Vector search, given the same question in words, returned unrelated passages at 0.49
similarity. That is not a better ranking — it is a question plain RAG **cannot express**.

**It helps when:**
- the link is a citation, not a similarity — the strongest case, as above
- the answer spans a chunk boundary — the `NEXT` edge alone fixes many of these
- you want fuller coverage of a topic, not just the single best passage (the HIV question
  pulls in more of the HIV rules via the shared `B20.0` code, even though they all happen
  to sit in the same section here)
- you need to *explain* a retrieval — every hit carries a `via`

**It does not help when:**
- one chunk fully answers the question — expansion just adds noise
- the graph is sparse (try `MAX_PAGES = 40` and watch it collapse into plain RAG)
- the extracted entities are wrong, in which case it is actively worse

**And it costs you:** a graph to build and store, two more parameters to tune
(`HOP_DECAY`, `MAX_HUB_DEGREE`), more context tokens per query, and a retrieval path that
is much harder to debug. The `via` field exists precisely because of that last one.

## Known weaknesses of what you built

**The code regex over-matches.** `A00` in "categories A00–T98" is a range boundary, not a
code mention, so a few edges are spurious.

**Section attribution is approximate.** A chunk is assigned to the last section that
started on or before its page. When two sections share a page, the second one's opening
chunks get mislabelled.

**One hop only.** `graph_search` never goes two hops out. Multi-hop reasoning
(*"the rule that the rule referenced by X points to"*) is out of reach.

**Hub suppression is a blunt instrument.** A fixed degree cutoff throws away genuinely
important high-frequency codes. Real systems weight by something like inverse document
frequency instead.

## Exercises, roughly in order of value

1. **Turn expansion off** — set `GRAPH_BUDGET = SEED_K`, re-run `head_to_head`. This is
   your control experiment. Any difference you saw was the graph.
2. **Break the graph deliberately** — set `MAX_PAGES = 40`, rebuild, and look at the edge
   counts. Very few codes repeat, so the graph has almost no cross-section links and
   Graph RAG converges on plain RAG. Sparse graph, no benefit.
3. **Tune `HOP_DECAY`.** At 0.9 neighbours rival direct matches and drown the answer; at
   0.1 they never make the cut. Find where it flips.
4. **Remove hub suppression** — set `MAX_HUB_DEGREE = 10000` and re-run the HIV question.
   Watch a single popular code drag in half the manual.
5. **Add two-hop expansion.** Wrap the expansion loop so neighbours can themselves expand
   (with a further decay). Does answer quality improve, or just the token count?
6. **Add a `SIBLING` edge** between chunks under the same *parent* section, so `4.2.1` and
   `4.2.3` connect. Does retrieval get better or noisier?
7. **Tighten `CODE_RE`** to reject range boundaries, then compare the edge count before
   and after. How many edges were noise?
8. **Try the LLM extraction cell** with `EXTRACT_N_CHUNKS = 3`. Read the raw output. This
   is the fastest way to understand why small models struggle with structured extraction.

## How this relates to the rest of the project

| | `rag-ouz` | `grag-ouz` (this) |
|---|---|---|
| Retrieval | top-k cosine similarity | seed by vector, expand by edges |
| Structure used | none | sections, codes, citations, reading order |
| Build time | seconds | seconds (structural) or hours (LLM) |
| Best at | direct lookup | aggregation, cross-reference, coverage |

The FastAPI service in `../app/` implements the `rag-ouz` approach. To move this one
across you would add a graph build step alongside `core/vectorstore.py` and extend
`core/rag.py` with the expansion logic — `graph_search()` is the function to port.
